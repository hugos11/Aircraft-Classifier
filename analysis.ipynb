```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import colormmaps
```

```python
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

# Data Setup from Experiment Log
versions = ['v1.0\n(Head-Only)', 'v2.0\n(Full Fine-Tune)', 'v3.0\n(Aug + 288px)', 'v4.0\n(RandAug + 384px)']

# Validation Accuracies
val_variant = [34.38, 80.80, 83.56, 86.35]
val_family = [44.28, 88.69, 91.15, 92.83]
val_manu = [56.74, 92.35, 95.14, 96.25]
train_acc_final = [52.22, 100.00, 98.89, 97.75]

# Epoch Training Loss Histories
epochs_v1 = list(range(1, 6))
loss_v1 = [3.6800, 2.7480, 2.3615, 2.0636, 1.8636]

epochs_10 = list(range(1, 11))
loss_v2 = [3.4842, 1.4667, 0.5470, 0.1955, 0.0779, 0.0364, 0.0217, 0.0170, 0.0146, 0.0136]
loss_v3 = [3.9455, 2.1682, 1.1416, 0.6448, 0.4040, 0.2764, 0.2100, 0.1699, 0.1438, 0.1422]
loss_v4 = [3.9498, 2.3384, 1.3370, 0.8213, 0.5541, 0.3917, 0.3047, 0.2503, 0.2213, 0.1943]
```

```python
w,x = 0.26, np.arange(len(versions))

fig, ax = plt.subplots(layout='constrained')
ax.bar(x-w, val_variant, width=w, label='Variant')
ax.bar(x, val_family, width=w, label='Family')
ax.bar(x+w, val_manu, width=w, label='Manufacturer')

ax.set_xticks(x)
ax.set_xticklabels(versions)
ax.set_ylabel('Percentage accuracy (%)')
ax.set_ylim(0,110)
ax.set_title('Fig. 1 - The progression in heirarchical accuracy over phases of development')
ax.legend(frameon = True)

plt.savefig('figures/fig_1_heirarchical_accuracy.png')
```

```python
fig, ax=plt.subplots(layout='constrained')

ax.plot(epochs_v1, loss_v1,ls='--', label='v1.0 (Head-Only), marker='D')
ax.plot(epochs_10, loss_v2,ls='--', label='v2.0(Full Fine-Tune)', marker='D')
ax.plot(epochs_10, loss_v3,ls='--', label='v3.0(Aug + 288px)', marker='D')
ax.plot(epochs_10, loss_v4,ls='--', label='v4.0 (RandAug + 384px)', marker='D')

ax.set_xticks(epochs_10)
ax.set_xlim(1,10)
ax.legend(frameon = True)
ax.set_xlabel('Epoch')
ax.set_ylabel('Training Loss (%))')
ax.set_title('Fig. 2 - Training Loss Convergence Dynamics Across Experiment Phases')

plt.savefig('figures/fig_2_training_loss.png')
```
